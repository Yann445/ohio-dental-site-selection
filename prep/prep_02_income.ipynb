## Ohio Dental Clinic — Site Selection Analysis

**Purpose:** Processes ACS 5-Year Estimates table B19013 (Median Household Income by ZIP Code Tabulation Area). Extracts and cleans median income for all 1,233 Ohio ZCTAs. Income is the primary revenue proxy in the composite scoring model (30% weight).

| | |
|---|---|
| **Input** | `B19013_income_raw.csv` (ACS 2020–2024) |
| **Output** | `B19013_income_cleaned.csv` |
| **Records** | 1,233 Ohio ZCTAs |

```python
# Import Libraries
import pandas as pd
import numpy as np
import os
import warnings
warnings.filterwarnings("ignore")

# PATHS
RAW_DATA_PATH = r"C:\Users\mosun\Downloads\oh_clinic_rw_files"
OUTPUT_PATH = r"../data/cleaned"

#  LOAD
print("Loading B19013_income_raw.csv...")
df = pd.read_csv(
    os.path.join(RAW_DATA_PATH, "ACSDT5Y2024.B19013-Data.csv"),
    header=0,       # row 1 = column codes → used as column names
    skiprows=[1],   # row 2 = human labels → skip it
    dtype=str,
    low_memory=False
)
print(f"Raw file: {len(df)} rows | {len(df.columns)} columns")

# Extract 5-digit ZIP from NAME column ("ZCTA5 43001" → "43001")
df["zip"] = df["NAME"].str.extract(r"(\d{5})")
df = df.dropna(subset=["zip"])
print(f"After ZIP extraction: {len(df)} rows")

# Convert income to numeric
# Census uses -666666666 for suppressed/unavailable values → replace with NaN
df["median_income"] = pd.to_numeric(df["B19013_001E"], errors="coerce")
df["median_income"] = df["median_income"].replace(-666666666, np.nan)
df["median_income"] = df["median_income"].replace(-999999999, np.nan)

# Report data quality
n_missing = df["median_income"].isna().sum()
print(f"ZIPs with suppressed/missing income: {n_missing} of {len(df)}")
print(f"Income range: ${df['median_income'].min():,.0f} — ${df['median_income'].max():,.0f}")
print(f"Median across all Ohio ZIPs: ${df['median_income'].median():,.0f}")

# Keep only the 2 columns needed
income_clean = df[["zip", "median_income"]].copy()

# SAVE
income_clean.to_csv(os.path.join(OUTPUT_PATH, "B19013_income_cleaned.csv"), index=False)
print(f"\nSaved: B19013_income_cleaned.csv")
print(f"Rows: {len(income_clean)} | Columns: {list(income_clean.columns)}")
print(f"\nSample:")
print(income_clean.head(5).to_string(index=False))
```
